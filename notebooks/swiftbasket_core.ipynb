```python
import os
import getpass

os.environ["SWIFTBASKET_DATABASE_URL"] = getpass.getpass(
    "Enter PostgreSQL connection URL: "
)

print("Database environment variable configured.")
```

```python
import swiftbasket_core

print("Gemini model:", swiftbasket_core.GEMINI_MODEL)
print("Database engine:", swiftbasket_core.engine.url.database)
print("Core module loaded successfully.")
```

```python
from swiftbasket_core import execute_sql_query

sql_test = """
SELECT AVG(total_order_value) AS average_order_value
FROM swiftbasket.orders;
"""

df = execute_sql_query(sql_test)

print(df)
```

```python
rag_test = swiftbasket_core.hybrid_retrieve(
    "Show me the details and status of order ORD0300000"
)

print("Retrieval type:", type(rag_test))
print("Number of results:", len(rag_test))

if rag_test:
    print("\nRetrieved document:")
    print(rag_test[0])
```

```python
rag_answer_test = swiftbasket_core.final_rag_answer(
    "Show me the details and status of order ORD0300000"
)

print("Retrieval Status:", rag_answer_test.get("retrieval_status"))
print("Retrieval Method:", rag_answer_test.get("retrieval_method"))
print("Exact Identifier:", rag_answer_test.get("exact_identifier"))
print("\nAnswer:")
print(rag_answer_test.get("answer"))
```

```python
result = swiftbasket_core.hybrid_answer(
    "What is the average order value?"
)

print("Route:", result.get("route"))
print("Status:", result.get("status"))
print("Generated SQL:", result.get("generated_sql"))
print("Answer/Data:")
print(result.get("data"))
```
